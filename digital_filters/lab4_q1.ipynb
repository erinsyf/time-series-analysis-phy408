**Question 1**

(a) The poles of the system are the singularities, given by denominator equal to zero: 

$$ \begin{cases} 
      z^{-1} - p = 0 &\Rightarrow z_{pole,1} = p^{-1} = \frac{e^{-i 2\pi f_0/f_s}}{1+\epsilon} \\
      \text{or} \\
      z^{-1} - p^* = 0 &\Rightarrow z_{pole,2} = (p^*)^{-1} = \frac{e^{i 2\pi f_0/f_s}}{1+\epsilon} 
   \end{cases}
$$

The zeros are given by the roots of the function, numerators equal to zero: 
$$ \begin{cases} 
      z^{-1} - q = 0 &\Rightarrow z_{zero,1} = q^{-1} = e^{-i 2\pi f_0/f_s} \\
      \text{or} \\
      z^{-1} - q^* = 0 &\Rightarrow z_{zero,2} = (q^*)^{-1} = e^{i 2\pi f_0/f_s}
   \end{cases}
$$

The unit circle in complex notation is given by numbers of the form $e^{i\theta}$. Note that for $0 < \epsilon \ll 1$, then the value of $1/(1+\epsilon) < 1$, hence the poles have magnitude less than 1, and thus lie inside of the unit circle (smaller than unit magnitude). Whereas the zeros lie on the unit circle. Since the poles have a magnitude less than one, this means the system is stable. Because we decompose the data in the $Z$ domain like the form $X(n) = \sum_n r^n e^{i \omega n} x(n)$ where $z^n = r^n e^{i \omega n}$, this means that only if  $r < 1$ this sum converges as $n \rightarrow \infty$. i.e., the sum converges if the magnitude of $z$ is less than unity. 

Additionally, the filter is real, because we have complex conjugate pairs of zeros and poles in both numerator and denominator. We can see this if we set $p = a + i b$ a general complex number, where $a, b \in \mathbb{R} $. Then the value of 
$$
\begin{align*} (z^{-1} - p)(z^{-1} - p^*) & =(z^{-1} - (a + ib))(z^{-1} - (a - ib)) \\ &= z^{-2} - (a+ib)z^{-1}-(a-ib)z^{-1} + (a+ib)(a-ib) = z^{-2} - 2az^{-1} + a^2 + b^2 \in \mathbb{R} 
\end{align*}
$$
Similarly with $q$ in the numerator. Hence the value of $W(z)$ is real. 

(b) Using the defined variables we can plot the absolute value squared of $W(f)$. Noting that for a range of frequency $f$ sampled between the Nyquist frequency, for 1000 points, the values of $z$ are given by $z = e^{i 2\pi f/f_s} = e^{i 2\pi f\Delta}$. The output plot is what we expect, the spectrum goes to zero at frequencies $f = \pm f_0 = \pm 1$. 

(c) To find the FWHM we can simply take one of the minimas (focus on positive frequencies), and find the indices where it was closest to where the value of the power spectrum reached half amplitude - 0.5 - on the left and right of the dip. Then taking the difference between these two frequency values we obtain $f_{fwhm} \approx 0.19$ cycles per year. The parameter we can change to make this value smaller (i.e. the dips skinnier) is to DECREASE $\epsilon$. This then makes the magnitude of the poles closer to 1, i.e. the amount of frequencies being rejected are less. 

(d) We can simply rewrite the equation like 
$$
\begin{align} W(z) &= M\frac{(z^{-1} - q)(z^{-1} - q^*)}{(z^{-1} - p)(z^{-1} - p^*)} \\ &= M\frac{z^{-2} - qz^{-1} - q^* z^{-1} + |q|^2}{z^{-2} - pz^{-1} - p^* z^{-1} + |p|^2} \\ \text{Notice } &\text{that for general complex number }\\ p + p^* &= a + i b + a - i b = 2a = 2\mathbb{R}(p) \\ \therefore &= M\frac{z^{-2} - 2\mathbb{R}(q)z^{-1} + |q|^2}{z^{-2} - 2\mathbb{R}(p)z^{-1} + |p|^2} \\ &= M\frac{(1/|p|^2)z^{-2} - (1/|p|^2) 2\mathbb{R}(q)z^{-1} + |q/p|^2}{(1/|p|^2)z^{-2} - (1/|p|^2)2\mathbb{R}(p)z^{-1} + 1} \\  &= \frac{a+bz^{-1} + cz^{-2}}{1+Fz^{-1} + Gz^{-2}} = \frac{N(z)}{D(z)}
\end{align}
$$
Hence comparing both sides we find that $a = M |q|^2/|p|^2$, $b = -2M \mathbb{R}(q)/|p|^2$, $c = M/|p|^2$, $F = -2\mathbb{R}(p)/|p|^2$ and $G = 1/|p|^2$. 

(e) Now we want to extend this result to a general filter that has coefficients $N = (n0, n1, n2, ...)$ and $D = (1, d1, d2, ...)$. For the filter we have above, we know that applying this filter in the $z$ domain means the response is of the form $Y(z) = W(z)X(z)$, and given $W(z) = N(z)/D(z)$, we get: 

$$
\begin{align*} D(z) Y(z) &= N(z) X(z) \\ (1+ Fz^{-1} G z^{-2}) Y(z) &= (a + b z^{-1} + c z^{-2}) X(z) \\ \text{By shift property } z^{-k} X(z) &= \sum_n x(n - k) z^{-n} \text{ and similarly for } Y(z) \\ \therefore \sum_n (y(n)z^{-n} + F y(n-1) z^{-n} + G y(n-2)z^{-n}) &= \sum_n (a x(n)z^{-n} + b x(n-1) z^{-n} + c x(n-2)z^{-n}) \\ \text{i.e. in code form at } &\text{each n, we sum up} \\ y[n] + Fy[n-1] + G y[n-2] & = a x[n] + b x[n-1] + c x[n-2] \\ \Rightarrow y[n] &= a x[n] + b x[n-1] + c x[n-2] - (Fy[n-1] + G y[n-2])
\end{align*}
$$
Where this includes two memory components with coefficients $F$ and $G$. So for general filter: 
$$ y[n] = \sum_{m = 0}^{len(N)} N[m] x[n - m] - \sum_{k = 1}^{len(G)} D[k] y[n - k]$$
We can simply just code this in two for loops. Note that for the loops to not fold in on themselves (when $n-m < 0$ or $n-k < 0$) we can just set a condition that only sums them if it's a valid positive index.

(f) Creating a delta impulse response, we can feed it into the ratFilter function written with the coefficients we found in part (d). From lecture, we know the delta response is a constant function in the $z$ domain - i.e. it has equal contribution from all frequencies. The output (from 0 to 6 years) shows a small response oscillation with frequency around 1 cycle/year (period of 1 year by observation). This makes sense because the notch filter aims to remove frequency of 1 cycle/year, and so the response generated from a delta response would be the exact oscillation needed to cancel it. The amplitude of the response decays with time, which is expected because we found that the poles have magnitude less than 1.

(g) Plotting both the Fourier transform of the impulse response, and the theoretical spectrum we found in part b), the observed spectrums are almost identical.

```python
## Question 1: Notch Filter 

# Import libraries 
import numpy as np 
import matplotlib.pyplot as plt
import scipy.fft as fft 

# Part b): Plot the power spectrum 

# Define variables 

f_s = 12 # cycles per year 
f_0 = 1 # cycles per year
M = 1.05 
eps = 0.05
delta = 1/f_s # Sampling interval (years)
q = np.exp(1j * 2* np.pi * f_0/f_s)
p = (1+ eps)* q

def W(f): 
    """
    Returns the value of the digital Notch filter
    Inputs: 
    f - frequency 
    Outputs: 
    Value of filter W
    """
    # By definition, where f * delta = Hz * period 
    z = np.exp(1j* 2*np.pi * f * delta) 
    w_z = M * (z**(-1) - q)*(
        z**(-1) - np.conjugate(q))/ ((z**(-1) - p)*(
        z**(-1) - np.conjugate(p)))
    return w_z

# Create sampling range of f 

f_sample = np.linspace(-f_s/2, f_s/2, 1000)

# calculate W
W_f = np.array([W(f) for f in f_sample])
# calculate power spectrum
power_spectrum = np.abs(W_f)**2 # ensure no imaginary rounding err

# plot it 
plt.plot(f_sample, power_spectrum)
plt.xlabel('Frequency f (cycles per year)')
plt.ylabel(r'Power Spectrum $|W(f)|^2$')
plt.title('Power Spectrum of Notch Filter')
plt.grid()
plt.show()

# Part c): Calculate FWHM

# focus on the positive side 
N = len(power_spectrum)
pos_spectrum = power_spectrum[N//2 : ]
pos_freqs = f_sample[N//2:]

# find minimum
i_peak = np.argmin(pos_spectrum)

# left side of peak, find closest index to half max = 0.5
left = np.argmin(np.abs(pos_spectrum[:i_peak] - 0.5))
print(f'left index = {pos_freqs[left]}')
# now right side
right = i_peak + np.argmin(np.abs(pos_spectrum[i_peak:] - 0.5))
print(f'right index = {pos_freqs[right]}')

fwhm = pos_freqs[right] - pos_freqs[left]

print(f'FWHM = {fwhm} cycles/year')

# Part e): Write ratFilter function

def ratFilter(N,D,x):
    """
    Function that returns filtered time series given 
    input x and coefficients of filter N and D
    inputs: 
    N - numerator coefficients [n0, n1,...]
    D - denominator coefficients [d0, d1, ...]
    x - input time series x[n]
    output: 
    Filtered time series using W(z) y[n]
    """
    # filtered output
    y = np.zeros_like(x, dtype=float)

    for n in range(len(x)):
        for m in range(len(N)): # for each coefficient
            y[n] += N[m] * x[n - m] 
        for k in range(1, len(D)): # starts from 1
            y[n] -= D[k] * y[n - k]
    return y

# Part f): Test impulse response
# Take 1200 samples, so dt = 1/fs
t = np.linspace(0, 100, 1200) # 100 years

# Notch filter coefficients
N = [M * np.abs(q)**2/np.abs(p)**2, -2*M * np.real(q)/np.abs(p)**2, M/np.abs(p)**2] 
D = [1, -2* np.real(p)/np.abs(p)**2, 1/np.abs(p)**2]

# Delta impulse input 
impulse = np.zeros_like(t, dtype = float) 
impulse[0] = 1.0

# Generate the response 
impulse_response = ratFilter(N, D, impulse)

# Plot response from 0 to 6 years 

plt.figure(figsize = (10,5))
plt.plot(t, impulse, label = 'Original Impulse')
plt.plot(t, impulse_response, label = 'Response')
plt.xlabel('Time (years)')
plt.ylabel('Amplitude')
plt.xlim(0, 6) # only plot 6 years
plt.title('Delta Impulse Response')
plt.legend()
plt.grid()
plt.show()

# Part g): Fourier Transform Impulse Response

impulse_fft = fft.fft(impulse_response)
freq = fft.fftfreq(len(t), d = delta)

# shift them to center zero 
impulse_fft = fft.fftshift(impulse_fft)
freq = fft.fftshift(freq)

plt.figure(figsize = (10, 5))
plt.plot(freq, np.abs(impulse_fft), label = 'Impulse FT')
plt.plot(f_sample, np.abs(W_f), color = 'red', linestyle = '--', label = 'Theoretical')
plt.xlabel('Frequency (cycles/year)')
plt.ylabel('Amplitude')
plt.legend()
plt.title('Frequency Spectrum of Impulse Response')
plt.grid()
plt.show()
```
